```python
from flask import Flask, request, jsonify
from datetime import datetime, timedelta

app = Flask(__name__)

# Weekly expiry days (Monday = 0 ... Sunday = 6)
expiry_days = {
    "NIFTY": 3,      # Thursday
    "BANKNIFTY": 2,  # Wednesday
    "FINNIFTY": 1    # Tuesday
}

def find_expiry(index, date):
    try:
        dt = datetime.strptime(date, "%Y-%m-%d")
    except:
        return None

    index = index.upper()
    if index not in expiry_days:
        return None

    target_day = expiry_days[index]
    diff = (target_day - dt.weekday()) % 7
    expiry_date = dt + timedelta(days=diff)

    return expiry_date.strftime("%Y-%m-%d")


# ---------------------- GET API ----------------------
@app.route("/expiry", methods=["GET"])
def get_expiry():
    index = request.args.get("index")
    date = request.args.get("date")

    if not index or not date:
        return jsonify({"error": "Use format: /expiry?index=NIFTY&date=2025-01-03"})

    expiry = find_expiry(index, date)
    if not expiry:
        return jsonify({"error": "Invalid index or date format"})

    return jsonify({"index": index.upper(), "date": date, "expiry": expiry})


# --


if __name__ == "__main__":
    app.run(debug=True, use_reloader=False)
```

```python
from flask import Flask, request, jsonify
from datetime import datetime, timedelta

app = Flask(__name__)

# Weekly expiry days (Monday = 0 ... Sunday = 6)
expiry_days = {
    "NIFTY": 3,      # Thursday
    "BANKNIFTY": 2,  # Wednesday
    "FINNIFTY": 1    # Tuesday
}

def find_expiry(index, date):
    try:
        dt = datetime.strptime(date, "%Y-%m-%d")
    except:
        return None

    index = index.upper()
    if index not in expiry_days:
        return None

    target_day = expiry_days[index]
    diff = (target_day - dt.weekday()) % 7
    expiry_date = dt + timedelta(days=diff)

    return expiry_date.strftime("%Y-%m-%d")


# ---------------------- POST API ----------------------
@app.route("/expiry", methods=["POST"])
def post_expiry():
    data = request.get_json()

    index = data.get("index")
    date = data.get("date")

    if not index or not date:
        return jsonify({"error": "Send JSON: {\"index\":\"NIFTY\",\"date\":\"2025-01-03\"}"})

    expiry = find_expiry(index, date)
    if not expiry:
        return jsonify({"error": "Invalid index or date format"})

    return jsonify({"index": index.upper(), "date": date, "expiry": expiry})


if __name__ == "__main__":
    app.run(debug=True, use_reloader=False,  port=5007)
```
